```python
# 1. Load Dataset 

# Import libraries 
import pandas as pd 
import matplotlib.pyplot as plt 

# Load dataset ('train.csv' from Kaggle House Prices Dataset)
df = pd.read_csv("train.csv")
```

```python
# 2. Explore 

# First 5 rows 
print("First 5 rows:") 
display(df.head()) # 5 rows are default 

# Data types 
print("\nData types of columns:")
print(df.dtypes) 
```

```python
# Clean Dataset 

# Check missing values
print("Missing values count per column:")
print (df.isnull().sum()) 

# Fix numeric conversions 
# Look for columns that should be numeric but are object 
for col in df.columns:
    if df[col].dtype == "object":
        try:
            df[col] = pd.to_numeric(df[col], errors="raise")
        except:
            # If conversion fails, leave it as object (it's categorical)
            pass

# Fill missing numeric values with mean 
numeric_cols = df.select_dtypes(include = ["int64", "float64"]).columns 
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean()) 

# Fill missing categorical values with mode 
categ_cols = df.select_dtypes(include = ["object"]).columns 
for col in categ_cols:
    df[col] = df[col].fillna(df[col].mode()[0]) # [0] means take first most common value 

# Show first 5 rows after cleaning 
print("\nFirst 10 rows after cleaning:")
print(df.head(10))

# Double-check that missing values are gone
print("\nRemaining missing values per column (should be 0):")
print(df.isnull().sum())
    
```

```python
# Basic Statistical Analysis 

# Select only numeric columns (int + float)
numeric_cols = df.select_dtypes(include=["int64","float64"]).columns 

# Loop through each numeric column 
for col in numeric_cols:
    print(f"\nColumn:{col}") # shows column name 
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode()[0])
    print("Variance:", df[col].var())
    print("Standard Deviation:", df[col].std()) 
```

```python
# Visualizations 

# Histogram for numeric column (SalePrice) 
plt.figure(figsize=(8,5))
plt.hist(df["SalePrice"], bins=30, edgecolor="black")
plt.title("Histogram of SalePrice")
plt.xlabel("Sale Price")
plt.ylabel("Number of Houses")
plt.show() 

# Bar chart for categorical column (Neighbourhood)
plt.figure(figsize=(12,5))
df["Neighborhood"].value_counts().plot(kind="bar")
plt.title("Number of Houses by Neighborhood")
plt.xlabel("Neighborhood")
plt.ylabel("Number of Houses")
plt.show()
```
