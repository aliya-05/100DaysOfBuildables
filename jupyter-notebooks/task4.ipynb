```python
import statistics 
import math 

# 1. Mean 
def mean(numbers):
    return statistics.mean(numbers)

# 2. Median 
def median(numbers):
    return statistics.median(numbers)

# 3. Mode 
def mode(numbers):
    try:
        return statistics.mode(numbers)
    except statistics.StatisticsError:
        return None # if all values appear once 

# 4. Variance 
def variance(numbers):
    return statistics.pvariance(numbers) # population variance (whole dataset)

# 5. Standard Deviation 
def std_dev(numbers):
    return statistics.pstdev(numbers) # population standard deviation (whole dataset) 

# 6. Euclidean Distance 
def euclidean_distance(x, y):
    if isinstance(x, (int, float)) and isinstance(y, (int, float)): 
        return abs(x - y) 
    elif isinstance(x, list) and isinstance(y, list): 
        return math.sqrt(sum((a - b) ** 2 for a, b in zip(x, y))) 
    else: 
        raise ValueError("Inputs must be both numbers or both lists") 

# 7. Sigmoid 
def sigmoid(x):
    return 1/(1 + math.exp(-x)) 
```

### What did I learn from these formulas?

- **Mean**: shows the average value of the data
- **Median**: gives the middle value and is less sensitive to outliers
- **Mode**: identifies the most repeated/common value in the dataset
- **Variance & Standard Deviation**: tells us how spread out the data set is
- **Euclidean Distance**: measures similarity/difference between numbers or lists
- **Sigmoid**: converts values into the range (0, 1), which is useful in probabilties and machine learning applications
