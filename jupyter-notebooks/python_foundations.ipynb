```python
# 1. Variables and Data Types 

# Variable: int
my_int = 14
print("Integer value:", my_int, "| Type:", type(my_int))

# Variable: float 
my_float = 3.14
print("Float value:", my_float, "| Type:", type(my_float))

# Variable: string 
text = "This is the first time I am using Jupyter Notebook!"
print("String value:", text, "| Type:", type(text))

# Variable: list 
numbers = [2, 4, 6, 8, 10]
print("Numbers list:", numbers, "| Type:", type(numbers))
strings = ["apple", "ball", "cat", "dog", "egg"]
print("Strings list:", strings, "| Type:", type(strings))
mixed = [2, "apple", 4, "ball", 6] 
print("Mixed list:", mixed, "| Type:", type(mixed))
```

```python
# 2. Conditionals 

user_num = int(input("Enter a number: ")) 

if user_num > 0:
    print("The number is positive.")
elif user_num < 0:
    print("The number is negative.")
else:
    print("The number is zero.")
```

```python
# 3. Loops 

# 3a. First 10 Natural Numbers 
for i in range(1, 11):
    print(i, end = " ")

# 3b. Factorial of a Number 
user_num = int(input("Enter a number to find it's factorial: "))
fact = 1
for i in range(1, user_num+1):
    fact *=i
print("Factorial of", user_num, "is", fact) 
```

```python
# 4. Functions 

# Function to greet a person 
def greet():
    name = input("Enter your name: ")
    print("Hello " + name + "!")

# Function to check if a number is even 
def is_even():
    user_num = int(input("Enter a number: "))
    if user_num % 2 == 0:
        print(user_num, "is even.")
    else:
        print(user_num, "is odd.") 

# Example Usage 
greet()
is_even()
```

```python
# 5. List Operations 

# Create a list of 5 numbers 
my_numbers = [1, 11, 14, 21, 23]

# Find max, min, sum and average 
maximum = max(my_numbers)
minimum = min(my_numbers)
total = sum(my_numbers)
average = total/len(my_numbers)

print("Numbers:", my_numbers)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Sum:", total)
print("Average:", average)

# Sort the list in ascending order 
my_numbers.sort()
print("Sorted list:", my_numbers)

# Reverse the list 
my_numbers.reverse()
print("Reversed list:", my_numbers) 
```

```python
# 5. List Operations with User Input 
# The part where I challenge myself 

# Create an empty list
my_numbers = []

# Take 5 numbers as input from user 
for i in range(5):
    num = int(input("Enter number " + str(i+1) + ": ")) 
    my_numbers.append(num)

# Find max, min, sum and average 
maximum = max(my_numbers)
minimum = min(my_numbers)
total = sum(my_numbers)
average = total/len(my_numbers)

print("\nNumbers:", my_numbers)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Sum:", total)
print("Average:", average) 

# Sort the list in ascending order
my_numbers.sort()
print("Sorted list:", my_numbers)

# Reverse the list
my_numbers.reverse()
print("Reversed list:", my_numbers) 
```
